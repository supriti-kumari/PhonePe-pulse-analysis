```python
# Put all the dataset in one dictionary

all_datasets = {
    "State_Txn and Users" : state_txn_users,
    "State_TxnSplit" : state_txn_split,
    "State_DeviceData":state_device_data,
    "District_Txn and Users":district_txn_users,
    "District Demographics":district_demo
}
```

```python
#Identify Missing Values

for name, df in all_datasets.items():

    print(f" Missing values in {name}")
    missing_count = df.isnull().sum()
    print(missing_count[missing_count >0] if missing_count.sum()>0 else "No missing values found")
    print()
```

```python
#Calculate the percentage of missing value in each column

for name, df in all_datasets.items():

    print(f" % Missing Values in {name}")

    missing_pct = (df.isnull().sum() / len(df))*100

    missing_pct = missing_pct[missing_pct>0]  #keep only columns that actually have missing values

    if len(missing_pct) == 0:
        print("No missing values found")

    else:
        print((missing_pct*100).round(2))

    print()
```

```python
# highlight which column has the highest percentage of missing values in each dataset

for name, df in all_datasets.items():

    missing_pct = (df.isnull().sum() / len(df))*100

    if missing_pct.sum() == 0:
        print(f" {name}: no missing values at all")

    else:

        worst_columns  = missing_pct.idxmax()   # column name with the highest % missing
        worst_value = missing_pct.max()

        print(f" {name} : worst column is '{worst_columns}' with {worst_value*100:.2f}% missing")
    
```
